# Blender Automation
I want to automate the process of importing data from OpenStreetMaps and saving it as an XML file.

Goal: Create a script that
* Takes a bounding box describing the simulation region
* Takes some parameters about which elements to import into the simulation, and what kind of vertex sampling to use
* Automatically exports this to an XML file using Mitsuba Blender
* Modifies the XML file to change the materials to useable formats by Sionna 2.0

This is going to use bpy to interface with the OpenStreetMaps and Mitsuba plugins
It assumes that we have a working installtion of Blender, with OpenStreetMaps and Mitsuba installed correctly

XML Modifications:
1. Dictionary of mappings from known elements to itu elements, then a default mapping
2. Swap the materials, and then downsize the material definitions so they don't repeat
3. I think this is everything? 

```python
import bpy
import addon_utils
import os
import sys

# Grabbing user preferences from the current Blender instance
bpy.ops.wm.read_userpref()

# Enabling addons and confirming
addon_utils.enable("mitsuba-blender")
addon_utils.enable("blosm")

# Enabling Blender in preferences
bpy.ops.preferences.addon_enable(module="mitsuba-blender")
bpy.ops.preferences.addon_enable(module="blosm")

# Adding Blosm to Blender path
addon_path = bpy.utils.user_resource("SCRIPTS")
blosm_path = os.path.join(addon_path, "blosm")
if blosm_path not in sys.path:
    sys.path.append(blosm_path)

if addon_utils.check("mitsuba-blender") == (True, True):
    print("Mitsuba Enabled")
if addon_utils.check("blosm") == (True, True):
    print("Blender OSM Enabled")
```

```python
def exportBoundingBox(bbox, elements, terrainReduction='1'):
    """
    Exports the XML scene from mitsuba blender from a specificed bounding box and detail parameters
    bbox: dict(str, float) the coordinates for the export
    elements: dict(str, float) the parts of the environment that you want to import
    terrainReduction: str ['1', '2', '5', '10', '1200']
    """

    # Deleting all existing objects
    for o in bpy.context.scene.objects:
        o.select_set(True)
    bpy.ops.object.delete()

    # Setting bounding box
    bpy.context.scene.blosm.minLat = bbox.get("minLat", 35.7198)
    bpy.context.scene.blosm.maxLat = bbox.get("maxLat", 35.7335)
    bpy.context.scene.blosm.minLon = bbox.get("minLon", -78.7038)
    bpy.context.scene.blosm.maxLon = bbox.get("maxLon", -78.6908)

    # Importing terrain first
    bpy.context.scene.blosm.dataType = 'terrain'
    bpy.context.scene.blosm.relativeToInitialImport = False
    terrain_options = ['1', '2', '5', '10', '1200']
    bpy.context.scene.blosm.terrainReductionRatio = terrainReduction if terrainReduction in terrain_options else '1'
    bpy.ops.object.shade_smooth()
    bpy.context.preferences.addons["blosm"].preferences.dataDir = "/home/everetttucker471/Documents/RL-AERPAW-DT/BlenderAutomation/temp_data/"
    bpy.ops.blosm.import_data()

    # Then import buildings
    bpy.context.scene.blosm.dataType = 'osm'
    # Setting scene elements, by default imports all
    bpy.context.scene.blosm.buildings = elements.get("buildings", True)
    bpy.context.scene.blosm.water = elements.get("water", True)
    bpy.context.scene.blosm.forests = elements.get("forests", True)
    bpy.context.scene.blosm.vegetation = elements.get("vegetation", True)
    bpy.context.scene.blosm.highways = elements.get("highways", True)
    bpy.context.scene.blosm.railways = elements.get("railways", True)

    # Additional Parameters
    bpy.context.scene.blosm.singleObject = True
    bpy.context.scene.blosm.relativeToInitialImport = False

    bpy.ops.blosm.import_data()

    # Now that all the data's in Blender, we need to export it using Mitsuba

    # Exporting as a .blend just to test if it works
    bpy.ops.wm.save_as_mainfile(filepath="/home/everetttucker471/Documents/RL-AERPAW-DT/BlenderAutomation/test_export/test-export.blend",
                                compress=True, check_existing=False)
    
    
    
exportBoundingBox({}, {}, '1')
    
```
